```python
import random
import matplotlib.pyplot as plt

def crearFortalezas():
    niveles = []
    nombres = ["Aetherion", "Drakmor", "Sylvaris", "Nocturna"]

    for nombre in nombres:
        while True:
            try:
                valor = int(input("Ingrese el nivel de criaturas en " + nombre + ": "))

                if valor < 0:
                    raise ValueError("No se permiten valores negativos")

                niveles.append(valor)
                break

            except ValueError as error:
                print("Error:", error)

    return niveles


def mostrarEstado(niveles):
    print("\n")
    print("AE:", niveles[0], "| DR:", niveles[1])
    print("")
    print("SY:", niveles[2], "| NO:", niveles[3])
    print("2")


def buscarFortaleza(nombre):

    fortalezas = ["aetherion", "drakmor", "sylvaris", "nocturna"]

    if nombre.lower() not in fortalezas:
        raise ValueError("La fortaleza no existe")

    return fortalezas.index(nombre.lower())


def defenderFortaleza(niveles):

    try:
        nombre = input("Ingrese la fortaleza a defender: ")

        indice = buscarFortaleza(nombre)

        niveles[indice] = niveles[indice] - 15

        if niveles[indice] < 0:
            niveles[indice] = 0

        print("Defensa realizada")

    except ValueError as error:
        print("Error:", error)

    mostrarEstado(niveles)


def pulsoArcano(niveles):

    for i in range(len(niveles)):
        niveles[i] = niveles[i] // 2

    print("Pulso Arcano activado")

    mostrarEstado(niveles)


def expansionOscura(niveles):

    indice = random.randint(0, 3)

    niveles[indice] += 10

    nombres = ["Aetherion", "Drakmor", "Sylvaris", "Nocturna"]

    print("La fortaleza afectada fue:", nombres[indice])

    mostrarEstado(niveles)


def ritualPurificacion(niveles):

    try:
        nombre = input("Ingrese la fortaleza a purificar: ")

        indice = buscarFortaleza(nombre)

        niveles[indice] = 0

        print("Purificación completada")

    except ValueError as error:
        print("Error:", error)

    mostrarEstado(niveles)


def desplazamientoCriaturas(niveles):

    try:

        origen = input("Fortaleza origen: ")
        destino = input("Fortaleza destino: ")

        indiceOrigen = buscarFortaleza(origen)
        indiceDestino = buscarFortaleza(destino)

        niveles[indiceDestino] += niveles[indiceOrigen]
        niveles[indiceOrigen] = 0

        print("Desplazamiento realizado")

    except ValueError as error:
        print("Error:", error)

    mostrarEstado(niveles)


def promedioAmenaza(niveles):

    acumulador = 0

    for valor in niveles:
        acumulador += valor

    promedio = acumulador / len(niveles)

    print("Promedio de amenaza:", promedio)

    if promedio > 25:
        print("¡Advertencia! El Reino Arcano está en peligro.")


def fortalezasPurificadas(niveles):

    nombres = ["Aetherion", "Drakmor", "Sylvaris", "Nocturna"]

    listaPurificadas = []

    for i in range(len(niveles)):

        if niveles[i] == 0:
            listaPurificadas.append(nombres[i])

    if len(listaPurificadas) == 0:
        print("No hay fortalezas purificadas")

    else:
        print("Fortalezas purificadas:")

        for nombre in listaPurificadas:
            print(nombre)


def reiniciarJuego():

    print("\nReiniciando juego...\n")

    return crearFortalezas()


def cargarDatos(niveles):

    try:

        archivo = open("datos.txt", "r")

        linea = archivo.readline()

        datos = linea.split(",")

        if len(datos) != 4:
            raise ValueError("El archivo debe contener 4 números")

        for i in range(4):
            niveles[i] = int(datos[i])

    except FileNotFoundError:
        print("No existe el archivo datos.txt")

    except ValueError as error:
        print("Error:", error)

    else:
        print("Datos cargados correctamente")

    finally:
        print("Proceso finalizado")

    mostrarEstado(niveles)


def mostrarGrafica(niveles):

    nombres = ["AE", "DR", "SY", "NO"]

    plt.figure(figsize=(8, 5))
    plt.bar(nombres, niveles)
    plt.title("Amenaza del Reino Arcano")
    plt.xlabel("Fortalezas")
    plt.ylabel("Criaturas")
    plt.show()


def mostrarTurnos(contadorTurnos):
    print("Turnos utilizados:", contadorTurnos)


def verificarVictoria(niveles):

    for valor in niveles:

        if valor != 0:
            return False

    return True


niveles = crearFortalezas()

contadorTurnos = 0

while True:

    mostrarEstado(niveles)

    print("\n===== MENÚ =====")
    print("1. Defender Fortaleza")
    print("2. Pulso Arcano")
    print("3. Expansión Oscura")
    print("4. Ritual de Purificación")
    print("5. Desplazamiento de Criaturas")
    print("6. Promedio de Amenaza")
    print("7. Fortalezas Purificadas")
    print("8. Reiniciar Juego")
    print("9. Cargar Datos")
    print("10. Mostrar Gráfica")
    print("11. Contador de Turnos")
    print("12. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        defenderFortaleza(niveles)
        contadorTurnos += 1

    elif opcion == "2":
        pulsoArcano(niveles)
        contadorTurnos += 1

    elif opcion == "3":
        expansionOscura(niveles)
        contadorTurnos += 1

    elif opcion == "4":
        ritualPurificacion(niveles)
        contadorTurnos += 1

    elif opcion == "5":
        desplazamientoCriaturas(niveles)
        contadorTurnos += 1

    elif opcion == "6":
        promedioAmenaza(niveles)
        contadorTurnos += 1

    elif opcion == "7":
        fortalezasPurificadas(niveles)
        contadorTurnos += 1

    elif opcion == "8":
        niveles = reiniciarJuego()
        contadorTurnos = 0

    elif opcion == "9":
        cargarDatos(niveles)
        contadorTurnos += 1

    elif opcion == "10":
        mostrarGrafica(niveles)
        contadorTurnos += 1

    elif opcion == "11":
        mostrarTurnos(contadorTurnos)

    elif opcion == "12":
        print("Gracias por jugar.")
        break

    else:
        print("Opción inválida")

    if verificarVictoria(niveles):
        print("\n¡Felicitaciones, has salvado el Reino Arcano y derrotado a las criaturas oscuras!")
        break
```
