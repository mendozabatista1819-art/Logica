```python
import matplotlib.pyplot as plt

#imaginemos que xes el tipo en años
x = [2000,2001,2002,2003,2004,2005]

#imaginemos que y la cantidad de cansiones
#lanzadas ese año
y = [5,2,0,8,3,5]

plt.bart(x,y)
plt.xlabel("años")
plt.xlabel("cansiones lanzadas")
plt.title("cansiones lanzadas por año")

plt.show()
```
